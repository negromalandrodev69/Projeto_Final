```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#1
#print inicial de leitura
bitcoin = pd.read_csv("investimentos.csv",sep=";")
display(bitcoin)
```

```python
# Classe das variaveis
display(bitcoin.info())
```

```python
#contagem de registros
id = bitcoin.groupby("id_registro").size()
print(id.to_string)
```

```python
#registro das datas
data = bitcoin.groupby("data_referencia").size()
bitcoin["data_referencia"] = pd.to_datetime(bitcoin["data_referencia"], format= "%Y-%m-%d",errors="coerce")

print(bitcoin["data_referencia"])
```

```python
#tipos de moedas
coin = bitcoin.groupby("moeda").size()
print(coin.to_string())
```

```python
#apreviação dos tipos diferentes de moeda
coin_a = bitcoin.groupby("codigo_moeda").size()
print(coin_a.to_string())
```

```python
# valores de compra
valor_c = bitcoin.groupby("cotacao_compra").size()
print(valor_c.to_string())
```

```python
# valores de venda
valor_v = bitcoin.groupby("cotacao_venda").size()
print(valor_v.to_string())
```

```python
# valor de variação das moedas
variavel = bitcoin.groupby("variacao_percentual").size()
print(variavel.to_string())
```

```python
# valor medio das moedas
media_c = bitcoin.groupby("cotacao_media").size()
print(media_c.to_string())
```

```python
# Nome do indicador se houver,em linhas vazias
indicador_c = bitcoin.groupby("indicador_economico").size()
print(indicador_c.to_string())
```

```python
#  O valor efetivo do índice naquela data,se houver taxa selic/ipca
indice = bitcoin.groupby("valor_indicador").size()
print(indice.to_string())
```

```python
# Tipo de porcentagem valor da taxa mensal/anual
taxa = bitcoin.groupby("unidade_medida").size()
print(taxa.to_string())
```

```python
# Frequencia na qual registra a cotação das moedas,podendo variar se houver taxas
foto = bitcoin.groupby("periodicidade").size()
print(foto.to_string())
```

```python
# Indica se a cotação é simulada ou real
fonte = bitcoin.groupby("fonte_dado").size()
print(fonte.to_string())
```
