```python
plt.figure(figsize=(30,20))
#maiores medias moedas
mmc = bitcoin.groupby("moeda")["cotacao_media"].max().sort_values(ascending=False)

mmc_bar = plt.bar(
    mmc.index,
    mmc.values,
    color=['r','g','b','y','#E900FF','#00EFFF','#39309C','#9BCA33','#EED243','#DD057F','#DD5905']
)

plt.title("COMPARATIVO DE MAIOR MEDIDA ENTRE MOEDAS",fontsize=25)
plt.ylim(0,50)

for coluna in mmc_bar:
    yvalor = coluna.get_height()
    plt.text(
        coluna.get_x() + coluna.get_width() / 2,
        yvalor,
        f'{yvalor:.2f}',
        fontsize=20,
        ha='center',
        va='bottom',
    )

plt.show()
```
