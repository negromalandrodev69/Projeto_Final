```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#1
#print inicial de leitura
bitcoin = pd.read_csv("investimentos.csv",dtype = {"data_referencia": str},sep=";")
display(bitcoin)
```

```python
#contagem de registros
id = bitcoin.groupby("id_registro").size()
display(id.to_string)
```

```python
#registro das datas
data = bitcoin.groupby("data_referencia").size()
print(data.to_string())
```

```python
#tipos de moedas
coin = bitcoin.groupby("moeda").size()
print(coin.to_string())
```

```python
#apreviação dos tipos diferentes de moeda
coin_a = bitcoin.groupby("codigo_moeda").size()
print(coin_a.to_string())
```

```python
# valores de compra
valor_c = bitcoin.groupby("cotacao_compra").size()
print(valor_c.to_string())
```

```python
# valores de venda
valor_v = bitcoin.groupby("cotacao_venda").size()
print(valor_v.to_string())
```

```python
# valor de variação das moedas
variavel = bitcoin.groupby("variacao_percentual").size()
print(variavel.to_string())
```
